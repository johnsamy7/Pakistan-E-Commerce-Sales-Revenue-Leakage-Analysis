```python
numeric_cols = ['price', 'qty_ordered', 'grand_total', 'discount_amount', 'MV']
```

```python
outlier_summary = []
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((df[col] < lower) | (df[col] > upper)).sum()
    pct = (count / len(df)) * 100

    outlier_summary.append({
        'Metric': col,
        'Min': df[col].min(),
        'Max': df[col].max(),
        'Lower Bound': round(lower, 2),
        'Upper Bound': round(upper, 2),
        'Outliers Count': count,
        'Outliers %': round(pct, 2)
    })

outlier_df = pd.DataFrame(outlier_summary)
print(outlier_df.to_string(index=False))
```

```python
negative_grand_total_count = (df['grand_total'] < 0).sum()
negative_discount_amount = (df['discount_amount'] < 0).sum()
prices_equal_to_zero = (df['price'] == 0).sum()
print(f"Number of negative values in 'grand_total': {negative_grand_total_count}")
print(f"Number of negative values in 'discount_amount': {negative_discount_amount}")
print(f"number of zeros in 'price': {prices_equal_to_zero}")
```

```python
REALIZED_STATUSES = ['complete', 'received', 'closed']
df_sales = df[df['status'].isin(REALIZED_STATUSES)].copy()

booked_revenue = df['grand_total'].sum()
realized_revenue = df_sales['grand_total'].sum()
leakage_pct = (1 - realized_revenue / booked_revenue) * 100
cancel_rate = (df['status'] == 'canceled').mean() * 100
refund_rate = df['status'].isin(['refund', 'order_refunded']).mean() * 100
mean_aov = df_sales['grand_total'].mean()
median_aov = df_sales['grand_total'].median()

print(f"Booked revenue:     {booked_revenue:,.0f}")
print(f"Realized revenue:   {realized_revenue:,.0f}")
print(f"Revenue leakage:    {leakage_pct:.1f}%")
print(f"Cancellation rate:  {cancel_rate:.1f}%")
print(f"Refund rate:        {refund_rate:.1f}%")
print(f"Mean AOV:           {mean_aov:,.0f}")
print(f"Median AOV:         {median_aov:,.0f}")
```

```python
payment_risk = df.groupby('payment_method').agg(
    Total_Orders=('increment_id', 'nunique'),
    Canceled_Orders=('status', lambda x: (x == 'canceled').sum())
).reset_index()

payment_risk = payment_risk[payment_risk['Total_Orders'] > payment_risk['Total_Orders'].median()]

payment_risk['Cancel_Rate_%'] = (payment_risk['Canceled_Orders'] / payment_risk['Total_Orders']) * 100

payment_risk = payment_risk.sort_values('Cancel_Rate_%', ascending=True)

payment_risk['Label'] = payment_risk.apply(
    lambda r: f"{r['Cancel_Rate_%']:.1f}%  (n={r['Total_Orders']:,.0f})", axis=1
)

fig = px.bar(
    payment_risk,
    x='Cancel_Rate_%',
    y='payment_method',
    orientation='h',
    color='Cancel_Rate_%',
    color_continuous_scale='Reds',
    text='Label',
    title='<b>Cancellation Rate by Payment Method</b><br><sup>Filtered for above-median order volume</sup>'
)

fig.update_layout(
    template='plotly_white',
    title_x=0.5,
    xaxis_title='',
    yaxis_title='',
    hovermode='closest',
    height=500,
    coloraxis_showscale=False,
    margin=dict(l=0, r=50, t=80, b=20)
)

fig.update_xaxes(
    showgrid=False,
    showticklabels=False,
    range=[0, payment_risk['Cancel_Rate_%'].max() * 1.25]
)

fig.update_yaxes(
    tickfont=dict(size=14),
    showspikes=False
)

fig.update_traces(
    textposition='outside',
    textfont_size=13,
    marker_line_width=0,
    width=0.6
)

fig.show()
```

```python
plot_data = df.copy()

plot_data['created_date'] = pd.to_datetime(plot_data['created_at'])
plot_data['join_date'] = pd.to_datetime(plot_data['Customer Since'])

plot_data['Order_M_Y'] = plot_data['created_date'].dt.to_period('M')
plot_data['Join_M_Y'] = plot_data['join_date'].dt.to_period('M')
plot_data['Customer_Type'] = (plot_data['Order_M_Y'] == plot_data['Join_M_Y']).map(
    {True: 'New Customer', False: 'Repeat Customer'}
)

monthly_cohort = plot_data.groupby(
    [pd.Grouper(key='created_date', freq='ME'), 'Customer_Type']
)['grand_total'].sum().reset_index()

fig = px.area(
    monthly_cohort,
    x='created_date',
    y='grand_total',
    color='Customer_Type',
    color_discrete_map={'New Customer': '#636efa', 'Repeat Customer': '#00cc96'},
    title='<b>Monthly Revenue Composition: New vs. Repeat Customers</b><br><sup>Visualizes retention health and acquisition sustainability</sup>',
    labels={'created_date': 'Date', 'grand_total': 'Monthly Revenue ($)', 'Customer_Type': 'Customer Segment'}
)

fig.update_layout(
    template='plotly_white',
    title_x=0.02,
    hovermode='x unified',
    height=550,
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1, title=""),
    margin=dict(l=40, r=30, t=100, b=40)
)

fig.update_yaxes(tickprefix="$", tickformat=",.0f", showgrid=True, gridcolor='rgba(0,0,0,0.05)')
fig.update_xaxes(showgrid=False)

fig.show()
```

```python
plot_df = df.copy()
plot_df['Payment_Channel'] = (
    plot_df['payment_method']
    .astype(str)
    .str.replace('_', ' ')
    .str.strip()
    .str.title()
)

# 2. Calculate Revenue & Share %
payment_summary = (
    plot_df.groupby('Payment_Channel')['grand_total']
    .sum()
    .reset_index(name='Revenue')
)
total_revenue = payment_summary['Revenue'].sum()
payment_summary['Share_%'] = (payment_summary['Revenue'] / total_revenue) * 100

# Keep top 8 channels and group the long tail into "Other"
if len(payment_summary) > 8:
    top_channels = payment_summary.nlargest(8, 'Revenue')
    other_revenue = payment_summary.loc[~payment_summary.index.isin(top_channels.index), 'Revenue'].sum()
    other_share = (other_revenue / total_revenue) * 100

    other_row = pd.DataFrame([{'Payment_Channel': 'Other Methods', 'Revenue': other_revenue, 'Share_%': other_share}])
    payment_summary = pd.concat([top_channels, other_row], ignore_index=True)

# Sort ascending so top earner is at the top
payment_summary = payment_summary.sort_values('Revenue', ascending=True)

# Create clear, direct labels: "$120,450 (45.2%)"
payment_summary['Label'] = payment_summary.apply(
    lambda r: f"${r['Revenue']:,.0f} ({r['Share_%']:.1f}%)", axis=1
)

# 3. Build the chart
fig = px.bar(
    payment_summary,
    x='Revenue',
    y='Payment_Channel',
    orientation='h',
    text='Label',
    title='<b>Revenue by Payment Method</b>',
    color_discrete_sequence=['#1f77b4']
)

# 4. Clean styling
fig.update_layout(
    template='plotly_white',
    title_x=0.02,
    xaxis_title='',
    yaxis_title='',
    height=450,
    margin=dict(l=20, r=120, t=60, b=20)
)

fig.update_xaxes(
    showgrid=False,
    showticklabels=False,
    range=[0, payment_summary['Revenue'].max() * 1.3]
)

fig.update_yaxes(
    tickfont=dict(size=13),
    showspikes=False
)

fig.update_traces(
    textposition='outside',
    textfont_size=12,
    width=0.6
)

fig.show()
```

```python
order_counts = (
    df.groupby('Customer ID')['increment_id']
    .nunique()
    .reset_index(name='Total_Orders')
)
order_counts['Order_Tier'] = pd.cut(
    order_counts['Total_Orders'],
    bins=[0, 1, 2, 5, float('inf')],
    labels=['1 Order (One-Time)', '2 Orders', '3–5 Orders', '6+ Orders (Loyal)']
)

loyalty_df = (
    order_counts['Order_Tier']
    .value_counts(sort=False)
    .reset_index()
)
loyalty_df.columns = ['Tier', 'Customers']

total_cust = loyalty_df['Customers'].sum()
loyalty_df['Share_%'] = (loyalty_df['Customers'] / total_cust) * 100
loyalty_df['Label'] = loyalty_df.apply(
    lambda r: f"{r['Customers']:,.0f} ({r['Share_%']:.1f}%)", axis=1
)

fig = px.bar(
    loyalty_df,
    x='Customers',
    y='Tier',
    orientation='h',
    text='Label',
    title='<b>Customer Loyalty Distribution</b>',
    color_discrete_sequence=['#2ca02c']
)
fig.update_layout(
    template='plotly_white',
    title_x=0.02,
    xaxis_title='',
    yaxis_title='',
    height=400,
    margin=dict(l=20, r=120, t=60, b=20)
)

fig.update_xaxes(
    showgrid=False,
    showticklabels=False,
    range=[0, loyalty_df['Customers'].max() * 1.3]
)

fig.update_yaxes(
    tickfont=dict(size=13),
    showspikes=False,
    autorange="reversed"
)
fig.update_traces(
    textposition='outside',
    textfont_size=12,
    width=0.6
)

fig.show()
```
